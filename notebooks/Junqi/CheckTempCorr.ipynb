```python
import cmocean.cm as cm
import matplotlib.pyplot as plt
import xarray as xr
```

```python
orig = xr.open_dataset('/data/sallen/results/MEOPAR/test_atmos/orig/SalishSea_1d_20230301_20230305_grid_T.nc')
corr = xr.open_dataset('/data/sallen/results/MEOPAR/test_atmos/SalishSea_1d_20230301_20230305_grid_T.nc')
```

```python
fig, axs = plt.subplots(1, 3, figsize=(15, 8))
orig.votemper[4, 0].plot(ax=axs[0], cmap=cm.thermal)
axs[0].set_title('Original, no correction')
corr.votemper[4, 0].plot(ax=axs[1], cmap=cm.thermal)
axs[1].set_title('With Temperature Correction')
(corr.votemper[4, 0] - orig.votemper[4, 0]).plot(ax=axs[2], cmap=cm.balance)
axs[2].set_title('Corrected - Original')
fig.suptitle('Difference on 5th Day');
```
